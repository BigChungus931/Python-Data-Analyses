# 2885. Rename Columns

```python
import pandas as pd
```

```python
student = {
    "id" : [1, 2, 3, 4, 5],
    "first" : ["Mason", "Ava", "Taylor", "Georgia", "Thomas"],
    "last" : ["King", "Wright", "Hall", "Thompson", "Moore"],
    "age" : [6, 7, 16, 18, 10]
}

students = pd.DataFrame(student)

students = students.rename(
    columns = {
        "id" : "student_id",
        "first" : "first_name",
        "last" : "last_name", 
        "age" : "age_in_years"
    }
)

students
```
