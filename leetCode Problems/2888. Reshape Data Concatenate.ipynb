# 2888. Reshape Data: Concatenate

```python
import pandas as pd
```

```python
dff1 = {
    "student_id" : [1, 2, 3, 4],
    "name" : ["Mason", "Ava", "Taylor", "Georgia"],
    "age" : [8, 6, 15, 17]
}

dff2 = {
    "student_id" : [5, 6],
    "name" : ["Leo", "Alex"],
    "age" : [7, 7]
}

df1 = pd.DataFrame(dff1)
df2 = pd.DataFrame(dff2)

students = pd.concat([df1, df2], ignore_index = True)

students
```
