# 2889. Reshape Data: Pivot

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "city": ["Jacksonville"]*5 + ["ElPaso"]*5,
    "month": ["January", "February", "March", "April", "May"]*2,
    "temperature": [13, 23, 38, 5, 34, 20, 6, 26, 2, 43]
})

result = df.pivot(index = "month", columns = "city", values = "temperature")

result
```
