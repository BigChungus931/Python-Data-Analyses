```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Datasets/personal data.csv")

df.describe()

df["mood"].value_counts()

df.head()
```

```python
avg_sleep = np.mean(df["sleep_hours"])

print(avg_sleep)

total_steps = np.sum(df["steps"])

print(total_steps)

highest_screen_time = df.loc[df["screen_time_hours"].idxmax()]

print(highest_screen_time)

pocket_money_total = np.sum(df["pocket_money_spent"])

print(pocket_money_total)
```

```python
plt.figure(figsize = (12, 6))

plt.plot(df["sleep_hours"])

plt.xlabel("Days")

plt.ylabel("Sleep Hours")

plt.title("Average sleep hours over days")

plt.show()
```

```python
fig, (ax, ax1) = plt.subplots(1, 2, figsize = (12, 6))

ax.bar(df["date"], df["steps"])

ax.tick_params(axis = "x", rotation = 90)

ax.set_xlabel("Days")

ax.set_ylabel("Steps")

ax.set_title("Steps on each day")

mood = df["mood"].value_counts()

ax1.pie(mood, labels = mood.index)

ax1.set_title("Mood over days")

ax1.legend(bbox_to_anchor=(0.2, 1.1))

plt.plot()
```
