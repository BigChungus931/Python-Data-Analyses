```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42) # Remove letter for randomness
steps = 1000

x = np.cumsum(np.random.choice([-1, 1], steps))
y = np.cumsum(np.random.choice([-1, 1], steps))

plt.figure(figsize=(6, 6))
plt.plot(x, y)
plt.title("Basic random walk")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.scatter(x, y, c=range(steps), cmap="inferno", s=5)
plt.axis("off")
plt.title("Colored random walk")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
for i in range(5):
    wx = np.cumsum(np.random.choice([-1, 1], steps))
    
    wy = np.cumsum(np.random.choice([-1, 1], steps))

    #plt.plot(wx, wy, alpha=0.7, label=f"Ant {i + 1}")
    plt.scatter(wx, wy, c=range(steps), cmap="plasma", s=5, label=f"Ant {i + 1}")

plt.legend()

plt.axis("off")

plt.title("Random race")

plt.show()
```

```python
results = []

for i in range(5):
    wx = np.cumsum(np.random.choice([-1, 1], steps))

    wy = np.cumsum(np.random.choice([-1, 1], steps))

    distance = np.sqrt(wx[-1]**2 + wy[-1]**2)

    results.append(
        {'ant' : f'ant {i + 1}', 'final_distance' : distance}
    )

df = pd.DataFrame(results)

leaderboard = df.sort_values('final_distance', ascending=False)

leaderboard
```

```python
bx = np.cumsum(np.random.choice([-1, 1], steps, p=[0.3, 0.7]))

by = np.cumsum(np.random.choice([-1, 1], steps, p=[0.3, 0.7]))

plt.figure(figsize=(8, 8))

plt.scatter(bx, by, c=range(steps), cmap="cool", s=5)

plt.axis("off")

plt.title("Biased random walk")

plt.show()
```
