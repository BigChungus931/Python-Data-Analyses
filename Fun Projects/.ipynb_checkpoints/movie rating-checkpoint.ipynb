# Movie Rating

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Datasets/movies.csv")
genre = df["genre"].value_counts()
print(genre)
df.head()
```

```python
avg_rating = np.mean(df["rating"])
print(avg_rating)

highest_rated = df.loc[df["rating"].idxmax()]

print(highest_rated)

lowest_rated = df.loc[df["rating"].idxmin()]

print(lowest_rated)

total_box_office = np.sum(df["box_office"])

print(total_box_office)

box_office_profit = df["box_office"] - df["budget"]

box_office_profit

top_5_movies = df.sort_values("rating", ascending=False).head(5)

print(top_5_movies[["title", "rating"]])

genre_avg_rating = df.groupby("genre")["rating"].mean().sort_values(ascending = False)
```

```python
plt.figure(figsize = (12, 6))

plt.bar(genre_avg_rating.index, genre_avg_rating.values, color="blue")

plt.xticks(rotation = 80)

plt.ylabel("Average Rating")

plt.title("Average Movie Rating By Genre")

plt.show()
```

```python
plt.figure(figsize = (12, 6))

plt.barh(top_5_movies["title"], top_5_movies["rating"], color="orange")

plt.xlabel("rating")

plt.title("Top 5 Highest Rated Movies")

plt.show()
```

```python
plt.figure(figsize = (12, 6))

plt.scatter(df["budget"], df["rating"], color="purple")

plt.xlabel("Budget (Millions)")

plt.ylabel("Rating")

plt.title("Budget vs Rating")

plt.show()
```
