# Football Teams Stats

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Datasets/seasons.csv")

print(df["result"].value_counts())

df.head()
```

```python
total_goals_scored = np.sum(df["goals_scored"])

print(f"Total goals scored: {total_goals_scored}")

total_goals_conceded = np.sum(df["goals_conceded"])

print(f"Total goals conceded: {total_goals_scored}")

avg_goal_per_game = np.mean(df["goals_scored"])

print(avg_goal_per_game)

home_away_stats = df.groupby("venue")[["goals_scored", "goals_conceded"]].mean()

home_away_stats
```

```python
plt.figure(figsize = (12, 6))

plt.plot(df["date"], df["goals_scored"], label="Goals scored", marker="o")

plt.plot(df["date"], df["goals_conceded"], label="Goals conceded", marker="o")

plt.xticks(rotation = 80)

plt.xlabel("Match date")

plt.ylabel("Goals")

plt.title("Goals scored vs Conceded over the season")

plt.legend()

plt.show()
```

```python
results_counts = df["result"].value_counts()

print(results_counts)

plt.bar(results_counts.index, results_counts.values, color=["green", "red", "grey"])

plt.title("Season results breakdown")

plt.ylabel("Number of matches")

plt.show()
```

```python
plt.pie(results_counts, labels = results_counts.index, autopct = "%1.1f%%")

plt.title("Win/Loss/Draw Ratio")

plt.show()
```
