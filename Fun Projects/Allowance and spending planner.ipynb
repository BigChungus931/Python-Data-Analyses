# Allowance and spending planner

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Datasets/budget.csv")
df
```

```python
total_money_spent = np.sum(df["amount"])

print(f"Total money spent: {total_money_spent}")

avg_spent = np.mean(df["amount"])

print(f"Average spent: {avg_spent}")

highest_category = df.loc[df["amount"].idxmax()]

print(f"Highest category: {highest_category}")

df["percentage"] = (df["amount"] / total_money_spent * 100)

df
```

```python
plt.figure(figsize = (8, 8))

plt.bar(df["category"], df["amount"], color = "mediumseagreen")

plt.xlabel("Category")

plt.ylabel("Amount Spent")

plt.title("Spending Category")

plt.show()
```

```python
plt.figure(figsize = (8, 8))

plt.pie(df["amount"], labels = df["category"], autopct = "%1.1f%%")

plt.title("Where my allowance goes")

plt.show()
```

```python
allowance = 135

savings_goal = 30

remaining_amount = allowance - total_money_spent

if remaining_amount >= savings_goal:
    print("You are on track hitting your savings goal")

else:
    print(f"You need to save {savings_goal - remaining_amount} more to hit your goal")
```
