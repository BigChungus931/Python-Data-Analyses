```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Datasets/weather.csv")
df.head()
#df.describe()
```

```python
mean_temp = np.mean(df["max_temp"])

print(mean_temp)

hottest_day = df.loc[df["max_temp"].idxmax()]

print(hottest_day)

total_rainfall = np.sum(df["rainfall"])

print(total_rainfall)

min_temp = df.loc[df["min_temp"].idxmin()]

print(min_temp)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["date"], df["max_temp"], label="Max temp")

plt.plot(df["date"], df["min_temp"], label="Min temp")

plt.xticks(rotation=80)

plt.xlabel("date")

plt.ylabel("temperature")

plt.title("Temp over month")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(df["date"], df["rainfall"], color="skyblue")

plt.xticks(rotation = 80)

plt.title("Daily rainfall")

plt.show()
```
