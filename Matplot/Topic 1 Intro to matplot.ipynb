# Intro to Matplot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4]

y = [10, 20, 15, 30]

plt.plot(x, y)

plt.show()
```

```python
fig, (ax, ax1, ax2) = plt.subplots(1, 3)

a = [1, 2, 3, 4]

b = [4000, 4300, 4100, 5100]

ax.plot(a, b)

ax.set_title("Monthly Revenue")

ax.set_xlabel("Months")

ax.set_ylabel("Revenue")

ax1.bar(["JAN", "FEB", "MAR", "APR"], b)

ax1.set_title("Bar Chart")

ax2.plot(b, a)
```

```python
x = np.linspace(0, 360, 100)
y = np.sin(np.radians(x))

plt.plot(x, y)

plt.axhline(0, color = "grey", linewidth = 1.2, linestyle = "--")

plt.xticks([0, 90, 180, 270, 360])

plt.yticks([-1, 0, 1])

plt.title("Sin Graph")

plt.ylabel("Sin(x)")

plt.xlabel("Degree")

plt.show()
```
