```python
import matplotlib.pyplot as plt
```

**Question 1**

```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x, y)

plt.title("Doubling Numbers")

plt.xlabel("x")

plt.ylabel("y")

plt.show()
```

**Question 2**

```python
fruits = ["apples", "bananas", "cherries"]

sales = [30, 45, 20]

plt.bar(fruits, sales, color=["red", "yellow", "darkred"])

plt.xlabel("Fruits")

plt.ylabel("Amount Sold")

plt.title("Sales of Fruits")

plt.show()
```

**Question 3**

```python
x = [1,2,3,4,5]

y = [5,3,4,2,6]

plt.scatter(x, y, marker="*", color="green")

plt.title("Scatter")

plt.xlabel("x")

plt.ylabel("y")

plt.show()
```

**Question 4**

```python
x = [1, 2, 3, 4, 5]

y = [5, 7, 2, 18, 22]

fig, (ax, ax1) = plt.subplots(1, 2, figsize=(10, 5))

ax.plot(x, y)

ax.set_title("Plot")

ax.set_xlabel("x")

ax.set_ylabel("y")

ax1.bar(x, y)

ax1.set_title("Bar")

ax1.set_xlabel("x")

ax1.set_ylabel("y")

plt.show()
```

**Question 5**

```python
y1=[1,4,9,16,25]

y2=[1,2,3,4,5]

x=[1,2,3,4,5]

plt.plot(x, y1, label="Line 1", color="blue")

plt.plot(x, y2, label="Line 2", color="red")

plt.title("Lines")

plt.xlabel("x")

plt.ylabel("y")

plt.legend()

plt.show()
```
