# Intro to Matplot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = [1, 2, 3, 4]

y = [10, 20, 15, 30]

plt.figure(figsize=(4, 3))

plt.plot(x, y)

plt.show()
```

```python
fig, (ax, ax1, ax2) = plt.subplots(1, 3, figsize=(15, 5))

a = [1, 2, 3, 4]

b = [4000, 4300, 4100, 5100]

ax.plot(a, b)

ax.set_title("Monthly Revenue")

ax.set_xlabel("Months")

ax.set_ylabel("Revenue")

ax1.bar(["JAN", "FEB", "MAR", "APR"], b)

ax1.set_title("Bar Chart")

ax2.plot(b, a)
```

```python
x = np.linspace(0, 360, 100)
y = np.sin(np.radians(x))

y1 = np.cos(np.radians(x))

plt.plot(x, y)

plt.axhline(0, color = "grey", linewidth = 1.2, linestyle = "--")

plt.xticks([0, 90, 180, 270, 360])

plt.yticks([-1, 0, 1])

plt.title("Sin Graph")

plt.ylabel("Sin(x)")

plt.xlabel("Degree")

plt.savefig("Images/Sin Graph.png")

plt.show()
```

```python
x = np.linspace(0, 360, 100)
y = np.sin(np.radians(x))

y1 = np.cos(np.radians(x))

plt.plot(x, y, label = "Sin x", color="blue")

plt.plot(x, y1, label = "Cos x", color="red")

plt.axhline(0, color = "grey", linewidth = 1.2, linestyle = "--")

plt.xticks([0, 90, 180, 270, 360])

plt.yticks([-1, 0, 1])

plt.title("Sin and Cos Graph")

plt.ylabel("Sin(x) and Cos(x)")

plt.xlabel("Degree")

plt.savefig("Images/Sin Graph.png")

plt.legend()

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30] 
y2 = [5, 8, 12, 9, 14]

# First Chart

plt.plot(x, y1)
plt.title("First Chart")
plt.savefig("Datasets/First Chart.png")

plt.clf()

# Second Chart

plt.plot(x, y2)
plt.title("Second Chart")
plt.savefig("Datasets/Second Chart.png")

plt.show()
```

```python
x = np.linspace(0, 360, 100)

y = np.cos(np.radians(x))

plt.plot(x, y)

plt.axhline(0, color = "grey", linewidth = 1.2, linestyle = "--")

plt.xticks([0, 90, 180, 270, 360])

plt.yticks([-1, 0, 1])

plt.title("Graph of Cos(x)")

plt.ylabel("Cos(x)")

plt.xlabel("Degree")

plt.show()
```

## Graphing a Pandas DataFrame

```python
df = pd.DataFrame({
    "Months": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [4000, 3000, 4300, 3900, 4100, 5200]
})

df

fig, ax = plt.subplots()

ax.plot(df["Months"], df["Revenue"], color = "red", marker = "o", linewidth = 1)

ax.set_title("Revenue by Month")

ax.set_ylabel("Revenue $")

ax.set_xlabel("Months")

ax.grid(True, color = "gray", linestyle = "--", linewidth = 0.5, alpha = 0.7)

plt.show()
```
