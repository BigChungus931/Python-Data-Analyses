# Real Data and Advance Styling

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Numpy: Generating and Transforming Data

```python
x = np.linspace(0, 10, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Sin x")
ax.text(3, 2, "y = sin(x)")
plt.show()
```

## Pandas: Plotting Straight From a DataFrame

```python
df = pd.read_excel("Datasets/students_clean.xlsx")
df.head()

fig, ax = plt.subplots()

df.groupby("grade")["gpa"].mean().plot(kind = "bar", ax = ax, color = "#4fd4ff")

ax.set_title("Average gpa by Grade")

ax.set_ylabel("gpa")

plt.show()
```

```python
fig, ax1 = plt.subplots()

ax1.hist(df["attendance_rate"], bins=10, color="#ff9b4f", label="Attendance %")

ax1.set_title("Attendance Rate Distribution")

ax1.set_xlabel("Attendance %")

ax1.legend()

plt.show()
```
