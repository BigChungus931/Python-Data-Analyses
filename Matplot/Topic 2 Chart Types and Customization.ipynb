# Chart Types and Customization

```python
import matplotlib.pyplot as plt
```

<img src = "Images/Charts.jpg">

```python
months = ["Jan", "Feb", "March", "April", "May"]

revenue = [4000, 2000, 3900, 4700, 5200]

fig, (ax, ax1) = plt.subplots(1, 2)

ax.plot(months, revenue)

ax.set_title("Monthly Revenue")

ax1.bar(months, revenue)

ax.set_title("Monthly Revenue")

plt.show()
```

```python
hours_studied = [1, 2, 3, 4, 5, 6]

exam_score = [52, 58, 60, 65, 74, 80]

fig, ax = plt.subplots()

ax.scatter(hours_studied, exam_score)

ax.set_title("Studied Time vs Exam Score")

plt.show()
```

```python
test_scores = [55, 62, 63, 70, 71, 72, 78, 80, 85, 91]

fig, ax = plt.subplots()

ax.hist(test_scores, bins = 10)

ax.set_xticks(test_scores)

ax.tick_params(axis = "x", rotation = 90)

plt.plot()
```
