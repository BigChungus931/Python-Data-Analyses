```python
import matplotlib.pyplot as plt
```

```python
# Question 1

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x, y)

plt.title("Doubling Numbers")

plt.xlabel("x")

plt.ylabel("y")

plt.show()
```

```python
# Question 2

fruits = ["apples", "bananas", "cherries"]

sales = [30, 45, 20]

plt.bar(fruits, sales, color=["red", "yellow", "darkred"])

plt.xlabel("Fruits")

plt.ylabel("Amount Sold")

plt.title("Sales of Fruits")

plt.show()
```
