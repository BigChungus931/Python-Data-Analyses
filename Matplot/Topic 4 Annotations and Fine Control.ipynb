# Annotations and Fine Control

```python
import matplotlib.pyplot as plt
from datetime import datetime
import matplotlib.dates as mdates
```

```python
month = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

revenue = [1000, 2000, 3500, 900, 5000, 1500]

fig, ax = plt.subplots()

ax.plot(month, revenue, marker="o")

ax.text(3, 4950, "Peak month", fontsize=10, color="green")

ax.set_xlabel("Months")

ax.set_ylabel("Revenue")

ax.set_title("Months against Revenue")

ax.annotate("Lowest month", xy=(3, 930), xytext=(1.6, 930), arrowprops=dict(arrowstyle="->", color="red"))

ax.annotate("Last month", xy=(5, 1500), xytext=(3.8, 1500), arrowprops=dict(arrowstyle="->", color="red"))

ax.annotate("Second peak", xy=(2, 3500), xytext=(0.7, 3500), arrowprops=dict(arrowstyle="->", color="red"))

plt.savefig("Datasets/Months against Revenue")

plt.show()
```

```python
dates = [datetime(2026, 1, 5), datetime(2026, 2, 12), datetime(2026, 3, 18), datetime(2026, 4, 2)]
revenue = [4000, 4300, 4100, 5100]

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(dates, revenue)

ax.xaxis.set_major_locator(mdates.MonthLocator())

ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))

ax.set_xlabel("Dates")

ax.set_ylabel("Revenue")

ax.set_title("Dates against Revenue")

fig.autofmt_xdate()

plt.savefig("Datasets/Dates against Revenue")

plt.show()
```
