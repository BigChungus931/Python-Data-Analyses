# Foundation and your first plots

<h2 style = "color:#4fd4ff" >What is Seaborn? - and how it sits on matplotlib</h2>

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
tips = pd.read_csv("Data Analysis/Datasets/tips.csv")
tips

sns.scatterplot(data = tips, x = "total_bill", y = "tip", hue = "time")
plt.show()
```

```python
sns.barplot(data = tips, x = "day", y = "total_bill")
plt.show()
```

```python
sns.histplot(data = tips, x = "total_bill", bins = 20, kde = True)
plt.show()
```

```python
flights = pd.read_csv("Data Analysis/Datasets/flights.csv")
flights

sns.lineplot(data = flights, x = "year", y = "passengers")
plt.show()
```

```python
sns.set_theme()
tipps = sns.load_dataset("tips")
sns.relplot(data = tipps, x = "total_bill", y = "tip", hue = "smoker", style = "smoker", size = "size", col = "time")
```

```python
dots = sns.load_dataset("dots")
dots.head(15)
```

```python
sns.relplot(data = dots, kind = "line", x = "time", y = "firing_rate", col = "align", hue = "choice", style = "choice", size = "coherence")
```

data=dots: “Use the table called dots.”

kind="line": “Draw lines.” You could ask for dots ("scatter") instead, but lines are better for showing change over time.

x="time": Time goes along the bottom of the graph.

y="firing_rate": How fast the brain cells fire goes up the side. A high line means very active cells.

col="align": “Make separate side-by-side graphs, one for each value in the align column.” That column has two values (the moment the dots appeared, and the moment the animal made its choice), so you get two small graphs next to each other. col means columns of graphs.

hue="choice": “Use different colors for different choices.” For example, one color for choice 1 and another for choice 2.

size="coherence": “Make the lines thicker or thinner depending on coherence.” Coherence is how many dots move the same way. Easy tasks (lots of dots moving together) and hard tasks get different line thickness.

style="choice": “Give each choice its own line style,” such as solid or dashed. Using color and style together makes the lines easier to tell apart, even in black and white.
