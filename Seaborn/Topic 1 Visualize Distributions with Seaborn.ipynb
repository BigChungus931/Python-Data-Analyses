# What is a distribution

A distribution shows how your data is spread out. <br>
It answers "which values show up a lot and which are rare".

### Part 1 - Histogram 

A histogram chops your data into groups called bins and counts how many items land in each bin. <br>
Each bar is one bin.

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(data=penguins, x="body_mass_g", bins=15, kde=True, color="#D6EFFF")
plt.title("How Heavy are penguins")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of penguins")
plt.show()
```

### Part 2 - The KDE Plot

KDE stands for **Kernel Density Estimate** is a fancy name for a smooth curve version of a histogram. <br>
Imagine it has a histogram where someone sanded off the blocky edges.

```python
sns.kdeplot(data=penguins, x="body_mass_g", fill=True, bw_adjust=0.2)
plt.show()
```

### Part 3 - Comparing Groups With Hue

If you want to compare the different species of penguins, we use the Hue to split data into different colors. <br>
From the graph below, we can quickly see that Gentoo penguins are the heaviest. <br>
If the bars overlap and are hard to read, we can set the attribute multiple="stack".

```python
sns.histplot(data=penguins, x="body_mass_g", hue="species", kde=True, multiple="stack")
plt.show()
```

### Part 4 - Making your own Data

We don't always need a dataset. let's pretend we have class' test score. <br>
This creates 200 scores centered around 75, usually within 10 points of it. 

```python
#np.random.seed(42) # Makes result the same everytime
scores = np.random.normal(loc=75, scale=10, size=200)
sns.histplot(scores, bins=20, kde=True, color="grey")
plt.show()
```

### Part 5 - The Boxplot

A boxplot squishes a distribution into a simple summary.

```python
sns.boxplot(data=penguins, x="species", y="body_mass_g")
plt.show()
```

### Part 6 - The Violin Plot

A violin plot is a boxplot and a kde plot combined. The wider the shape, the more data is there.

```python
sns.violinplot(data=penguins, x="species", y="body_mass_g")
plt.show()
```
